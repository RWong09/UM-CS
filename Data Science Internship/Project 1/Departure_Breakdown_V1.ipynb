##### 1. Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

plt.rcParams.update({
    'figure.dpi': 130,
    'axes.spines.top': False,
    'axes.spines.right': False,
    'axes.titlesize': 13,
    'axes.titleweight': 'bold'
})
```

##### 2. Load Dataset

```python
df = pd.read_csv('DEPARTURE_BREAKDOWN_V1.csv')
print(f'Shape: {df.shape}')
df.head()
```

##### 3. Check for Missing Data

```python
missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)

missing_report = pd.DataFrame({
    'Missing Count': missing,
    'Missing (%)': missing_pct
})

print('--- Missing Data Report ---')
print(missing_report.to_string())
print()

total_missing = missing.sum()
print(f'Total missing values across all columns: {total_missing}')
if total_missing == 0:
    print('✅ No missing data found across all 15 columns. Proceeding to preprocessing.')
else:
    print(f'⚠️  Total missing values: {total_missing}')
```
